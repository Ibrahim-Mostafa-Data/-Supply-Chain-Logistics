```python
# ـOutliers Check
outlier_cols = [
    'Sales per customer',
    'Benefit per order',
    'Order Item Discount',
    'Order Item Product Price',
    'Order Item Total',
    'Product Price',
    'Days for shipping (real)',
    'Days for shipment (scheduled)'
]
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q1 - Q3 
    lower = Q1 - 1.5 * IQR
    upper = Q3 - 1.5 * IQR
    outlier = ((df[col]< lower)|(df[col]>upper)).sum()
print(col, outlier)
```

```python
df['Customer Lname'] = df['Customer Lname'].fillna('Unknown')
```

```python
df['Customer Zipcode'] = df['Customer Zipcode'].replace('nan', float('nan'))
```

```python
# Convert to datetime
df['shipping date (DateOrders)'] = pd.to_datetime(df['shipping date (DateOrders)'])
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])

# Get invalid date cases
invalid_dates = df[
    df['shipping date (DateOrders)'] < df['order date (DateOrders)']
]

# Calculate difference in days
date_diff = (
    invalid_dates['shipping date (DateOrders)']
    - invalid_dates['order date (DateOrders)']
).dt.days

print('Minimum:', date_diff.min())
print('Maximum:', date_diff.max())
print('Mean:', date_diff.mean())
```
